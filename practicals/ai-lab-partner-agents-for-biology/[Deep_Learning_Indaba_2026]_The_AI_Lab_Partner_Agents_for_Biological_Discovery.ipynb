### 🔍 2.1 Discovery Prompt

**Prompt:**
> How does the sequence determine the thermostability of a protein? Please give me a scientifically correct description of the different mechanisms involved while avoiding heavy jargon where possible


<details>
<summary>🤖 Show Example Output</summary>

At its core, a protein’s sequence determines its thermostability by dictates how the 3D folded structure balances two opposing forces: the energy holding the protein together (enthalpy) and the natural tendency of molecules to become disorganized (entropy).

At high temperatures, thermal energy makes the atoms vibrate violently. If these vibrations overcome the forces holding the structure together, the protein unfolds (denatures).

Here are the key sequence-dependent mechanisms that nature uses to make proteins heat-resistant:

1. **Tight Packing of the "Water-Hating" (Hydrophobic) Core**
   - **The Concept:** Most proteins have a water-avoiding core. Amino acids with hydrophobic side chains (like Leucine, Isoleucine, and Valine) cluster together inside the protein, away from water.
   - **Mechanism:** Thermostable proteins have sequences optimized to pack these hydrophobic amino acids together with almost no gaps or cavities. This maximizes weak but numerous physical forces (van der Waals interactions) that act like velcro, holding the core tightly together.

2. **Surface Charge Networks (Salt Bridges)**
   - **The Concept:** Opposites attract. Positively charged amino acids (like Arginine and Lysine) attract negatively charged ones (like Glutamic acid and Aspartic acid).
   - **Mechanism:** In heat-resistant proteins, these charged amino acids are frequently arranged on the outer surface of the protein in cooperative networks (called salt bridges). Because water molecules behave differently at high temperatures, these surface electrostatic networks actually become stronger or remain highly stable, holding the outer shell of the protein together.

3. **Covalent Staples (Disulfide Bonds)**
   - **The Concept:** Covalent bonds are the strongest chemical bonds in a protein.
   - **Mechanism:** The sequence can place Cysteine amino acids in positions where they form covalent links (disulfide bridges) with each other. These act like physical staples. When the protein is heated, these staples physically restrict the chain from unraveling, drastically reducing the structural paths it can take to unfold.

4. **Rigidifying Flexible Loops**
   - **The Concept:** Proteins are not rigid blocks; they have flexible "loop" regions connecting structured helices and sheets. Thermal unfolding almost always starts at these flexible loops.
   - **Mechanism:**
     - **Proline (The Rigidifier):** Proline is an amino acid with a rigid, ring-like structure that naturally restricts the movement of the protein backbone. Replacing loop amino acids with Proline limits how much the loop can move, preventing unfolding.
     - **Glycine (The Flexer):** Glycine has no side chain and is highly flexible. Heat-resistant proteins often have fewer Glycine residues in their loop regions to prevent them from wild whipping motions at high temperatures.

5. **Enhanced Hydrogen Bonding**
   - **The Concept:** Hydrogen bonds are weak chemical interactions that form between polar atoms.
   - **Mechanism:** Thermostable proteins typically have a higher density of hydrogen bonds throughout their structure. By carefully positioning polar amino acids (like Serine, Threonine, or Asparagine), the sequence ensures that if one hydrogen bond breaks due to heat, many others are nearby to immediately hold the structure in place.

**Summary**

If you are designing a sequence to be more thermostable, you are generally trying to:
- Fill internal cavities with hydrophobic residues.
- Add charge partnerships (salt bridges) on the surface.
- Staple the structure with disulfide bonds.
- Replace flexible loop residues (less Glycine, more Proline) to restrict movement.

</details>

Here is a quick summary of why targeting surface residues is the perfect starting point for a beginner in protein engineering:

- **High Forgiveness, Low Risk**: The inside core of a protein is extremely fragile; one wrong mutation there will usually cause the whole protein to collapse or misfold. The surface is highly forgiving, allowing beginners to experiment with mutations without instantly destroying the protein's overall 3D shape.
- **Simple, Intuitive Mechanisms**: The most common surface engineering trick is adding "Molecular Velcro" (salt bridges). It is conceptually very easy for a beginner to understand: you just look for two amino acids near each other on the surface and mutate them into opposite electrical charges so they snap together.
- **Easy to Calculate**: Calculating surface exposure is computationally cheap and easy to visualize. Modeling the precise thermodynamic packing of the internal core requires much heavier, more advanced physics simulations.

It serves as the perfect "sandbox" to learn the computational tools before moving on to the high-risk, high-reward challenge of engineering the protein's core!

```python
# @title Code for visualising surface-exposed protein regions

import os
import urllib.request
import py3Dmol
from Bio.PDB import PDBParser
from Bio.PDB.SASA import ShrakeRupley

# Theoretical maximum SASA (Å²) for the 20 standard amino acids (Tien et al. 2013)
TIEN_MAX_ASA = {
    'ALA': 129.0, 'ARG': 274.0, 'ASN': 195.0, 'ASP': 193.0, 'CYS': 167.0,
    'GLU': 223.0, 'GLN': 225.0, 'GLY': 104.0, 'HIS': 224.0, 'ILE': 197.0,
    'LEU': 201.0, 'LYS': 236.0, 'MET': 224.0, 'PHE': 240.0, 'PRO': 159.0,
    'SER': 155.0, 'THR': 172.0, 'TRP': 285.0, 'TYR': 263.0, 'VAL': 174.0
}

def get_surface_exposed_residues(pdb_filepath, rsa_threshold=0.20):
    """Calculates the Relative Solvent Accessibility (RSA) for all residues in a PDB file.
    Identifies residues with RSA higher than rsa_threshold as surface-exposed.
    """
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure("protein", pdb_filepath)
    # Compute absolute SASA at the residue level
    sr = ShrakeRupley(probe_radius=1.4, n_points=100)
    sr.compute(structure, level="R")
    surface_residues = []
    core_residues = []
    for model in structure:
        for chain in model:
            chain_id = chain.id
            for residue in chain:
                # Filter out water, ligands (HETATM) and non-standard residues
                if residue.get_id()[0] == ' ':
                    resname = residue.get_resname()
                    resnum = residue.get_id()[1]
                    if resname in TIEN_MAX_ASA:
                        abs_sasa = residue.sasa
                        max_sasa = TIEN_MAX_ASA[resname]
                        rsa = abs_sasa / max_sasa
                        residue_data = {
                            "chain": chain_id,
                            "resnum": resnum,
                            "resname": resname,
                            "abs_sasa": abs_sasa,
                            "rsa": rsa
                        }
                        if rsa >= rsa_threshold:
                            surface_residues.append(residue_data)
                        else:
                            core_residues.append(residue_data)
    return surface_residues, core_residues

# --- Download and Analyze Insulin (PDB: 1MSO) ---
pdb_id = "1MSO"
pdb_filename = f"{pdb_id}.pdb"
pdb_url = f"https://files.rcsb.org/download/{pdb_filename}"

# Download PDB file if not already present
if not os.path.exists(pdb_filename):
    print(f"Downloading {pdb_filename}...")
    urllib.request.urlretrieve(pdb_url, pdb_filename)

# Run the analysis (Threshold set to 20% RSA)
surface_exposed, core = get_surface_exposed_residues(pdb_filename, rsa_threshold=0.20)
print(f"\nAnalysis complete:")
print(f"Total surface-exposed residues (RSA >= 20%): {len(surface_exposed)}")
print(f"Total core residues (RSA < 20%): {len(core)}")

# --- Visualize using py3Dmol ---
with open(pdb_filename, 'r') as f:
    pdb_data = f.read()

viewer = py3Dmol.view(data=pdb_data, format="pdb")
viewer.setBackgroundColor('white')
# Style all residues to a neutral gray/light-blue default (Core representation)
viewer.setStyle({'cartoon': {'color': '#E5E9F0'}})

# Highlight surface-exposed residues in a contrasting Magenta/Orange color
for res in surface_exposed:
    viewer.setStyle(
        {'chain': res['chain'], 'resi': res['resnum']},
        {'cartoon': {'color': '#FF5733'}} # Contrasting Orange-Red
    )

# Retain the representation of critical disulfide bonds as yellow sticks
viewer.addStyle({'resn': 'CYS'}, {'stick': {'colorspec': 'yellowCarbon', 'radius': 0.12}})
viewer.addStyle({'elem': 'S'}, {'sphere': {'color': 'yellow', 'radius': 0.35}})
viewer.zoomTo()
viewer.show()

# Print a subset of the surface residues as potential mutation targets
print("\nSample Surface-Exposed Residues (Potential Mutation Targets):")
print(f"{'Chain':<6}{'Res Num':<8}{'Res Name':<10}{'RSA (Relative Solvent Accessibility, %)':<8}")
for res in sorted(surface_exposed, key=lambda x: (x['chain'], x['resnum']))[:10]:
    print(f"{res['chain']:<6}{res['resnum']:<8}{res['resname']:<10}{res['rsa']*100:.1f}%")
```

```python
# @title Code for making mutations and visualising change in structure

import os
import requests
import py3Dmol
import Bio.PDB
from Bio.PDB import PDBParser
from Bio.PDB.SASA import ShrakeRupley

# Theoretical maximum SASA (Tien et al. 2013)
TIEN_MAX_ASA = {
    'ALA': 129.0, 'ARG': 274.0, 'ASN': 195.0, 'ASP': 193.0, 'CYS': 167.0,
    'GLU': 223.0, 'GLN': 225.0, 'GLY': 104.0, 'HIS': 224.0, 'ILE': 197.0,
    'LEU': 201.0, 'LYS': 236.0, 'MET': 224.0, 'PHE': 240.0, 'PRO': 159.0,
    'SER': 155.0, 'THR': 172.0, 'TRP': 285.0, 'TYR': 263.0, 'VAL': 174.0
}

# --- Helper Functions ---
def fold_with_esmfold(sequence: str, name: str) -> str:
    """Folds sequence using the correct public ESMFold API."""
    clean_seq = "".join(sequence.split())
    url = "https://api.esmatlas.com/foldSequence/v1/pdb/"
    headers = {"Content-Type": "text/plain"}
    print(f"Folding {name} ({len(clean_seq)} AA) via ESMFold...")
    response = requests.post(url, data=clean_seq, headers=headers)
    if response.status_code == 200:
        return response.text
    else:
        raise Exception(f"ESMFold failed (Status {response.status_code}): {response.text}")

def get_surface_residues(pdb_filepath, rsa_threshold=0.20):
    """Calculates RSA and returns surface-exposed residues (excluding CYS)."""
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure("protein", pdb_filepath)
    sr = ShrakeRupley(probe_radius=1.4, n_points=100)
    sr.compute(structure, level="R")
    surface_residues = []
    for model in structure:
        for chain in model:
            for residue in chain:
                if residue.get_id()[0] == ' ':
                    resname = residue.get_resname()
                    resnum = residue.get_id()[1]
                    # Exclude Cysteine (CYS) to avoid breaking disulfide bridges
                    if resname in TIEN_MAX_ASA and resname != "CYS":
                        rsa = residue.sasa / TIEN_MAX_ASA[resname]
                        if rsa >= rsa_threshold:
                            surface_residues.append({
                                "resnum": resnum,
                                "resname": resname,
                                "rsa": rsa
                            })
    # Sort by highest exposure
    surface_residues.sort(key=lambda x: x['rsa'], reverse=True)
    return surface_residues

def align_structures(ref_pdb_path, sample_pdb_path, output_pdb_path):
    """Aligns sample structure to reference structure using C-alpha atoms."""
    parser = PDBParser(QUIET=True)
    ref_struct = parser.get_structure("ref", ref_pdb_path)
    sample_struct = parser.get_structure("sample", sample_pdb_path)
    ref_atoms = []
    sample_atoms = []
    for ref_res, sample_res in zip(ref_struct[0].get_residues(), sample_struct[0].get_residues(), strict=True):
        if ref_res.get_id()[0] == ' ' and sample_res.get_id()[0] == ' ':
            if 'CA' in ref_res and 'CA' in sample_res:
                ref_atoms.append(ref_res['CA'])
                sample_atoms.append(sample_res['CA'])
    super_imposer = Bio.PDB.Superimposer()
    super_imposer.set_atoms(ref_atoms, sample_atoms)
    super_imposer.apply(sample_struct[0].get_atoms())
    io = Bio.PDB.PDBIO()
    io.set_structure(sample_struct)
    io.save(output_pdb_path)
    return super_imposer.rms

# --- Execution Pipeline ---
# 1. Wild-type Single-Chain Insulin Sequence
wt_sequence = (
    "FVNQHLCGSHLVEALYLVCGERGFFYTPKT"  # B-chain (1-30)
    "GGGPRR"                          # Linker  (31-36)
    "GIVEQCCTSICSLYQLENYCN"           # A-chain (37-57)
)

# 2. Fold WT and save PDB
wt_pdb_text = fold_with_esmfold(wt_sequence, "Wild-Type")
wt_pdb_file = "insulin_wt.pdb"
with open(wt_pdb_file, "w") as f:
    f.write(wt_pdb_text)

# 3. Analyze surface residues of WT
surface_candidates = get_surface_residues(wt_pdb_file)

# 4. Select top 2 surface residues to mutate to Proline (P)
mutations_to_make = surface_candidates[:2]
print(f"\nSelected mutations based on highest surface exposure (RSA):")
for mut in mutations_to_make:
    print(f"- Position {mut['resnum']}: {mut['resname']} (RSA: {mut['rsa']*100:.1f}%) -> PRO")

# 5. Generate mutant sequence string
seq_list = list(wt_sequence)
for mut in mutations_to_make:
    idx = mut['resnum'] - 1  # 0-based index
    seq_list[idx] = 'P'
mut_sequence = "".join(seq_list)
print(f"Mutated Sequence: {mut_sequence}")

# 6. Fold mutant and save PDB
mut_pdb_text = fold_with_esmfold(mut_sequence, "Mutant")
mut_pdb_file = "insulin_mut_raw.pdb"
with open(mut_pdb_file, "w") as f:
    f.write(mut_pdb_text)

# 7. Align Mutant coordinates to WT coordinates to allow visual comparison
aligned_mut_file = "insulin_mut_aligned.pdb"
rmsd = align_structures(wt_pdb_file, mut_pdb_file, aligned_mut_file)
print(f"\nStructural alignment completed. RMSD: {rmsd:.3f} Å")
if rmsd < 1.5:
    print("Fold is successfully conserved (RMSD is low).")
else:
    print("Significant structural rearrangement detected (RMSD is high).")

# 8. Visualise the Overlay in py3Dmol
with open(wt_pdb_file, "r") as f:
    wt_data = f.read()
with open(aligned_mut_file, "r") as f:
    mut_data = f.read()

viewer = py3Dmol.view(width=600, height=500)
viewer.setBackgroundColor('white')
# Model 0: Wild-type (Colored Green)
viewer.addModel(wt_data, "pdb")
viewer.setStyle({'model': 0}, {'cartoon': {'color': '#A3BE8C', 'opacity': 0.85}})
# Model 1: Aligned Mutant (Colored Salmon Red)
viewer.addModel(mut_data, "pdb")
viewer.setStyle({'model': 1}, {'cartoon': {'color': '#BF616A', 'opacity': 0.85}})

# Highlight the mutation sites as spheres on both models
for mut in mutations_to_make:
    resnum = mut['resnum']
    # WT residue in green spheres
    viewer.addStyle({'model': 0, 'resi': resnum}, {'sphere': {'color': '#3B7A57', 'radius': 0.8}})
    # Mutated residue (Proline) in red spheres
    viewer.addStyle({'model': 1, 'resi': resnum}, {'sphere': {'color': '#8B0000', 'radius': 0.8}})

viewer.zoomTo()
viewer.show()
```

> Is there a quantitative way to measure how these mutations affect the stability of the protein? What are the strengths and weaknesses of the best models for this task, and are there any blind spots I should be cautious of


<details>
<summary>🤖 Show Example Output</summary>

**Measuring Stability Quantitatively: ΔΔG**

Yes, there is a strict thermodynamic metric for this. It is called ΔΔG (Delta Delta G), which represents the change in Gibbs Free Energy.

- ΔG is the energy difference between a protein being neatly folded versus unraveled like spaghetti.
- ΔΔG is the difference in that energy between your Wild-Type and your Mutant.
- The Rule: A negative ΔΔG (e.g., -1.5 kcal/mol) means the mutation is stabilizing. A positive ΔΔG means it is destabilizing.

**The Best Models for Predicting ΔΔG**

There are three main categories of tools you can use to calculate this, each with unique strengths and weaknesses:

1. **Physics-Based Force Fields (e.g., Rosetta cartesian_ddg)**
- Strengths: This is the gold standard. Rosetta physically simulates the atomic forces (electrostatics, van der Waals forces, hydrogen bonds) to calculate how much energy it takes to pack the new mutation into the space. It physically rotates the sidechains to find the lowest energy fit.
- Weaknesses: It is computationally massive and extremely slow. You generally need a high-performance computing cluster, not just a Colab notebook.

2. **Empirical Fast Calculators (e.g., FoldX)**
- Strengths: FoldX uses simplified mathematical shortcuts derived from real physical experiments. It is lightning-fast and can scan hundreds of mutations in minutes.
- Weaknesses: It is incredibly sensitive to the exact starting coordinates of the atoms. If your starting PDB file has even a slight clash (two atoms drawn slightly too close together), FoldX will throw a massive error penalty and give you wildly inaccurate results.

3. **Machine Learning Models (e.g., RaSP, MutateX)**
- Strengths: These are the newest generation. Models like RaSP use 3D convolutional neural networks to look at the pocket around the mutation and instantly predict the Rosetta-level ΔΔG in a fraction of a second.
- Weaknesses: Like all AI, they are limited by their training data. If your protein has a highly unusual fold or unusual metals bound to it, the AI might hallucinate an incorrect stability score.


**⚠️ The Blind Spots (Crucial Warnings)**

If you are entering this field, there are two massive blind spots you must avoid:

- **Blind Spot 1: "The AlphaFold Illusion"**

The most common mistake beginners make is assuming that because AlphaFold or ESMFold predicted a beautiful, high-confidence structure for a mutant, the mutant must be stable. This is false. AlphaFold is essentially a highly advanced pattern-recognition engine for finding the final 3D shape; it is terrible at thermodynamics. You can introduce a mutation that completely destroys the protein's hydrophobic core (making it wildly unstable in real life), and AlphaFold will often still confidently draw it perfectly folded. You must use a dedicated ΔΔG tool (like FoldX or RaSP) to prove stability, not a folding algorithm.

- **Blind Spot 2: Stability vs. Aggregation**

You might design a mutation that gives you a fantastic, highly negative ΔΔG. The protein is technically more stable. However, if your mutation exposed a sticky, hydrophobic patch on the surface, the proteins will clump together into a useless sludge (aggregation) when you try to manufacture them in the lab. A protein must be both thermodynamically stable and highly soluble!

```python
# @title 2. Define thermal stability scoring function
def get_thermal_score(sequence: str) -> float:
  """Predicts optimal growth temperature (OGT) in °C using ThermoFormer."""
  truncate_length = 2048
  if len(sequence) > truncate_length:
    sequence = sequence[:truncate_length]
  inputs = tf_tokenizer(
      [sequence], return_tensors='pt', padding=True, truncation=True
  )
  input_ids = inputs['input_ids'].to(device)
  attention_mask = inputs['attention_mask'].to(device)
  with torch.no_grad():
    outputs = tf_model(input_ids, attention_mask=attention_mask)
  return outputs.predicted_values.cpu().numpy().tolist()[0]
```

```python
# @title 3. Define single-position mutation scoring function (ESM-2)
def score_mutation(sequence: str, pos: int, wild_type: str, mutant: str) -> float:
  """Scores a mutation at `pos` (0-indexed) using the pseudo-log-likelihood ratio."""
  assert 0 <= pos < len(sequence), 'Position out of bounds.'
  assert sequence[pos] == wild_type, f'Expected {wild_type} at index {pos}, but found {sequence[pos]}'

  # Mask the target position
  seq_list = list(sequence)
  seq_list[pos] = tokenizer.mask_token
  masked_sequence = ''.join(seq_list)

  # Tokenize and perform inference
  inputs = tokenizer(masked_sequence, return_tensors='pt').to(device)
  with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits  # [batch_size, seq_len, vocab_size]

  # Map sequence position to token position (handles <cls> start token)
  token_pos = 1 + pos

  # Get token IDs
  wt_id = tokenizer.convert_tokens_to_ids(wild_type)
  mut_id = tokenizer.convert_tokens_to_ids(mutant)

  # Extract log probabilities
  log_probs = torch.log_softmax(logits[0, token_pos], dim=-1)

  # Calculate score
  score = log_probs[mut_id] - log_probs[wt_id]
  return score.item()
```

## 3.3 Building a General Sequence Scoring Function

### 🔍 Discovery Prompt

**Prompt:**
> How does ESM-2 evaluate and score a whole sequence?


<details>
<summary>🤖 Show Example Output</summary>

ESM-2 outputs logit scores over the entire vocabulary for every sequence position. To score a whole sequence rather than a single-point mutation delta, we compute the average log-probability of all actual residues in the sequence under the model.

Mathematically:
$$ \text{Fitness Score} = \frac{1}{L} \sum_{i=1}^L \log p(x_i \mid X) $$
where $p(x_i \mid X)$ is the model's predicted probability of the correct amino acid $x_i$ at position $i$ given the unmasked sequence context $X$.
Excluded from the average calculation are special tokens like `<cls>`, `<eos>`, and `<pad>`.

Biologically, a higher average log-probability score indicates that the sequence contains amino acid patterns that are more evolutionary favorable or consistent with the natural proteins ESM-2 was trained on. A mutation that increases this score suggests overall stabilization or evolutionary compatibility.

> **Student Note on Special Tokens:**
> - `<cls>` (classification) is a helper token at the start of a sequence that pools global representation.
> - `<eos>` (end-of-sequence) marks the end.
> - `<pad>` (padding) is used to pad sequences to equal length when batching.
> These helper tokens are introduced by the tokenizer and do not represent real biology. We exclude them from calculations to prevent skewing the average log-probability.

### Implementation Prompt

**Prompt:**
> The single-position score tells me about one mutation at a time.
> Can you write a function that scores an entire protein sequence
> using ESM-2 — something like an overall evolutionary fitness
> score for the whole sequence? I want to compare wild-type vs
> mutant sequences.

</details>



```python
# @title Sequence fitness score function

def get_sequence_fitness_score(sequence: str, model, tokenizer, device) -> float:
  """Calculates the fitness score of a protein sequence using ESM-2.

  The score is computed as the average log-probability of the actual tokens
  present in the sequence, excluding special tokens (<cls>, <eos>, <pad>, etc.).
  """
  # 1. Tokenize the unmasked sequence
  inputs = tokenizer(sequence, return_tensors='pt')
  inputs = {k: v.to(device) for k, v in inputs.items()}
  input_ids = inputs['input_ids']

  # 2. Perform forward pass
  with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits  # Shape: [1, seq_len + 2, vocab_size]

  # 3. Compute log-probabilities over vocabulary at each position
  log_probs = torch.log_softmax(logits, dim=-1)

  # 4. Retrieve the log-probability of the actual token present at each position
  seq_log_probs = torch.gather(log_probs, dim=-1, index=input_ids.unsqueeze(-1)).squeeze(-1)

  # 5. Exclude special tokens from the average
  special_ids_tensor = torch.tensor(tokenizer.all_special_ids, device=device)
  special_tokens_mask = torch.isin(input_ids, special_ids_tensor)
  non_special_mask = ~special_tokens_mask

  # Extract only non-special tokens log-probabilities
  valid_log_probs = seq_log_probs[non_special_mask]

  # Compute average log-probability across all residues
  average_log_prob = valid_log_probs.mean().item()

  return average_log_prob
```

## 3.5 Attempt 1: Random Mutation (Baseline)

### 🔍 Discovery Prompt

**Prompt:**
> What is the simplest way to search for mutations that optimize this score, and what are its limits?


<details>
<summary>🤖 Show Example Output</summary>

The simplest strategy for searching sequence space is a randomized hill-climbing search (specifically, **First-Choice Hill Climbing**). At each step:
1. Propose a single random amino acid substitution at a random position.
2. Score the candidate sequence using our oracle.
3. Accept the mutation only if it strictly improves the fitness score compared to the current sequence.

**Limitations:**
- **Low Efficiency:** Most random mutations are neutral or deleterious (harmful to stability/function). Thus, the search takes many steps to find even a minor improvement.
- **No Biological Focus:** It explores positions completely at random without considering structural context (e.g. active sites, surface vs core).
- **Epistasis:** It only makes single steps and can easily get stuck in local optima. Epistasis means the effect of a mutation at position A depends on position B. If moving to a higher peak requires traversing a fitness valley (where both mutations are needed together), a single-point random search will reject the individual mutations along the way and get stuck.

```python
# @title Code for making random mutations
import random

def random_mutate(sequence: str) -> tuple[str, int, str, str]:
  """Randomly selects a position and mutates it to a different amino acid.

  Args:
    sequence: The input protein sequence.

  Returns:
    A tuple of (mutated_sequence, position, wild_type_aa, mutant_aa).
    Position is 0-indexed.
  """
  amino_acids = 'ACDEFGHIKLMNPQRSTVWY'
  pos = random.randint(0, len(sequence) - 1)
  wt_aa = sequence[pos]

  # Choose a different AA
  possible_mutants = [aa for aa in amino_acids if aa != wt_aa]
  mutant_aa = random.choice(possible_mutants)

  mutated_sequence = sequence[:pos] + mutant_aa + sequence[pos+1:]
  return mutated_sequence, pos, wt_aa, mutant_aa

# Optimization loop (5 steps)
current_seq = wt_sequence
current_score = get_sequence_fitness_score(current_seq, model, tokenizer, device)

print(f"Initial Sequence: {current_seq}")
print(f"Initial Score: {current_score:.6f}\n")

for step in range(1, 6):
  candidate_seq, pos, wt_aa, mutant_aa = random_mutate(current_seq)
  candidate_score = get_sequence_fitness_score(candidate_seq, model, tokenizer, device)

  accepted = False
  if candidate_score > current_score:
    current_seq = candidate_seq
    current_score = candidate_score
    accepted = True

  mutation_str = f"{wt_aa}{pos + 1}{mutant_aa}"
  print(f"Step {step}:")
  print(f"  Proposed Mutation: {mutation_str}")
  print(f"  Candidate Score:   {candidate_score:.6f}")
  print(f"  Accepted:          {accepted}")
  print(f"  Current Best Score: {current_score:.6f}\n")
```

```python
# @title Code for getting mutation suggestions
import json
import re
from google import genai
from google.genai import types
from google.colab import userdata
from pydantic import BaseModel, Field

# Helper to apply a mutation string (e.g. 'R35A') to a sequence
def apply_mutation(sequence: str, mutation: str) -> str:
  if not mutation:
    return sequence
  try:
    wt_aa = mutation[0]
    pos = int(mutation[1:-1]) - 1  # 0-indexed
    mut_aa = mutation[-1]

    seq_list = list(sequence)
    seq_list[pos] = mut_aa
    return "".join(seq_list)
  except Exception as e:
    print(f"Error applying mutation '{mutation}': {e}")
    return sequence

# Configure Gemini API client
try:
  api_key = userdata.get('GEMINI_API_KEY')
  client = genai.Client(api_key=api_key)
  HAS_GEMINI = True
except Exception as e:
  HAS_GEMINI = False
  print("Warning: GEMINI_API_KEY not found in Colab Secrets. Gemini steps will be skipped.")

class MutationSuggestion(BaseModel):
  mutation: str = Field(
      description="The proposed single-point mutation, e.g., 'R35P' (WT residue, 1-based position, mutant residue)"
  )
  reasoning: str = Field(
      description="A brief explanation of the biophysical rationale (e.g. packing, charge, hydrogen bonds, literature context)"
  )

def ask_gemini_for_mutation(
    sequence: str,
    current_score: float,
    tried_mutations: list[dict] = None,
) -> tuple[str, str]:
  """Prompts Gemini to suggest a single point mutation in the sequence."""
  if not HAS_GEMINI:
    return "", "Gemini API key not configured."

  history_str = ""
  if tried_mutations:
    filtered_history = [item for item in tried_mutations if item.get('step', 0) > 0]
    if filtered_history:
      history_str = "\n  Previously tried mutations:\n"
      for item in filtered_history:
        status = "ACCEPTED" if item.get('accepted') else "REJECTED"
        thermal_val = item.get('thermal') if item.get('thermal') is not None else item.get('candidate_thermal')
        score_info = f"Thermal: {thermal_val:.2f} °C" if thermal_val is not None else "Thermal: N/A"
        if 'composite_score' in item:
          score_info += f", Composite Score: {item.get('composite_score'):.4f}"
        history_str += f"  - {item['mutation']}: {status} ({score_info})\n"
        history_str += f"    Reasoning used: {item.get('reasoning')}\n"
      history_str += "\n  CRITICAL: Do NOT propose any of the previously tried mutations listed above. You must suggest a new single-point mutation that is not in the history.\n"

  prompt = f"""
  You are an expert protein engineer. Propose a single point mutation in the following protein sequence to improve its thermal stability.

  Protein Sequence (Human Single-Chain Insulin):
  {sequence}

  Current Thermal Stability Score (predicted OGT): {current_score:.2f} °C
  {history_str}
  Propose exactly ONE mutation. Use biological reasoning (e.g., sequence conservation, secondary structure propensities, hydrophobic packing, or introducing disulfide bonds/salt bridges).
  """

  try:
    response = client.models.generate_content(
        model='gemini-3.1-flash-lite',
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=MutationSuggestion,
            temperature=1.0,
        ),
    )
    suggestion: MutationSuggestion = response.parsed
    return suggestion.mutation.strip(), suggestion.reasoning.strip()
  except Exception as e:
    return "", f"Error querying Gemini: {e}"

# --- Compare Random vs Gemini-guided mutation ---

# 1. Base WT scores
wt_thermal_score = get_thermal_score(wt_sequence)
wt_esm_score = get_sequence_fitness_score(wt_sequence, model, tokenizer, device)

# 2. Random mutation
rand_mut = random_mutate(wt_sequence)  # returns (mutated_seq, pos, wt_aa, mut_aa)
rand_seq, rand_pos, rand_wt_aa, rand_mut_aa = rand_mut
rand_mut_str = f"{rand_wt_aa}{rand_pos + 1}{rand_mut_aa}"
rand_thermal = get_thermal_score(rand_seq)
rand_esm = get_sequence_fitness_score(rand_seq, model, tokenizer, device)

# 3. Gemini mutation
gemini_mut_str, gemini_reason = "", "Gemini API key not set"
gemini_thermal, gemini_esm = 0.0, 0.0
if HAS_GEMINI:
  print("Querying Gemini for guided mutation...")
  gemini_mut_str, gemini_reason = ask_gemini_for_mutation(wt_sequence, wt_thermal_score)
  if gemini_mut_str:
    gemini_seq = apply_mutation(wt_sequence, gemini_mut_str)
    gemini_thermal = get_thermal_score(gemini_seq)
    gemini_esm = get_sequence_fitness_score(gemini_seq, model, tokenizer, device)

# 4. Print comparison
print("\n" + "=" * 110)
print(f"{'Mutation Type':<20} | {'Mutation':<10} | {'Thermal (°C)':<14} | {'Δ Thermal':<12} | {'ESM-2':<12} | {'Reasoning'}")
print("=" * 110)
print(f"{'Wild-Type':<20} | {'-':<10} | {wt_thermal_score:<14.2f} | {0.0:<12.2f} | {wt_esm_score:<12.6f} | {'Baseline'}")
print(f"{'Random':<20} | {rand_mut_str:<10} | {rand_thermal:<14.2f} | {rand_thermal - wt_thermal_score:<+12.2f} | {rand_esm:<12.6f} | {'Random amino acid substitution'}")
if gemini_mut_str:
  print(f"{'Gemini-Guided':<20} | {gemini_mut_str:<10} | {gemini_thermal:<14.2f} | {gemini_thermal - wt_thermal_score:<+12.2f} | {gemini_esm:<12.6f} | {gemini_reason}")
else:
  print(f"{'Gemini-Guided':<20} | {'N/A':<10} | {'N/A':<14} | {'N/A':<12} | {'N/A':<12} | {gemini_reason}")
print("=" * 110)
```

### Implementation Prompt

<details>
<summary>🤖 Note on Loop Memory </summary>

> A simple memoryless loop has a major limitation: if Gemini proposes a mutation
> that gets rejected because OGT didn't improve, it has no record of that failure
> and may propose the same rejected mutation again. In a production-grade pipeline,
> we maintain a history of previously tried mutations (both accepted and rejected)
> and append them to the Gemini prompt context to guide it away from failed attempts.
</details>


```python
# @title Code for an optimisation loop
def optimize_protein(initial_sequence: str, steps: int = 5) -> tuple[str, list[dict]]:
  """Runs a closed-loop optimization using Gemini to propose mutations and ThermoFormer as the oracle."""
  current_sequence = initial_sequence
  current_thermal = get_thermal_score(current_sequence)

  history = []
  print(f"Starting closed-loop optimization...")
  print(f"Initial Sequence: {current_sequence}")
  print(f"Initial Thermal Score: {current_thermal:.2f} °C\n")

  for step in range(1, steps + 1):
    print(f"--- Step {step} ---")

    # 1. Ask Gemini for mutation (passing history memory)
    if HAS_GEMINI:
      mutation_str, reasoning = ask_gemini_for_mutation(current_sequence, current_thermal, history)
    else:
      print("Gemini API not available. Skipping step.")
      break

    if not mutation_str:
      print("No mutation proposed or error occurred. Skipping step.")
      continue

    # 2. Apply proposed mutation
    candidate_sequence = apply_mutation(current_sequence, mutation_str)

    # 3. Score candidate sequence with ThermoFormer
    candidate_thermal = get_thermal_score(candidate_sequence)

    # 4. Accept if thermal score strictly improves
    accepted = False
    if candidate_thermal > current_thermal:
      current_sequence = candidate_sequence
      print(f"Mutation {mutation_str} ACCEPTED (Thermal: {current_thermal:.2f} -> {candidate_thermal:.2f} °C)")
      current_thermal = candidate_thermal
      accepted = True
    else:
      print(f"Mutation {mutation_str} REJECTED (Thermal: {candidate_thermal:.2f} <= current best {current_thermal:.2f} °C)")

    print(f"Reasoning: {reasoning}")
    print(f"Current Best Thermal Score: {current_thermal:.2f} °C\n")

    history.append({
        'step': step,
        'mutation': mutation_str,
        'thermal': candidate_thermal,
        'accepted': accepted,
        'reasoning': reasoning,
        'sequence': current_sequence,
        'thermal_score': current_thermal,
    })

  return current_sequence, history

# Execute the optimization loop for 5 steps starting with wild-type sequence
optimized_sequence, optimization_history = optimize_protein(wt_sequence, steps=5)

# Find and print the best sequence details
accepted_history = [h for h in optimization_history if h['accepted']]
if accepted_history:
  best_entry = accepted_history[-1]
  print("\n" + "="*50)
  print("OPTIMIZATION SUMMARY: BEST SEQUENCE FOUND")
  print("="*50)
  print(f"Step:            {best_entry['step']}")
  print(f"Mutation:        {best_entry['mutation']}")
  print(f"Thermal OGT:     {best_entry['thermal']:.2f} °C")
  print(f"Reasoning:       {best_entry['reasoning']}")
  print("="*50)
  print(f"Sequence:\n{optimized_sequence}")
  print("="*50)
else:
  print("\nNo stabilizing mutations were accepted. The wild-type remains the best sequence.")
```

## 3.8 🧠 [Optional / Advanced] Function Preservation

### 🔍 Discovery Prompt

**Prompt:**
> How do I measure if mutations preserve the protein's original function?


<details>
<summary>🤖 Show Example Output</summary>

To ensure our optimization loop doesn't trade biological function for raw thermostability, we can calculate sequence and semantic similarity metrics relative to the wild-type template:

1. **Edit Distance (Levenshtein Distance):**
   - **Concept:** Calculates the minimum number of single-character edits (insertions, deletions, substitutions) to transform the mutated sequence back to wild-type.
   - **Limitation:** It is a raw sequence metric that penalizes all substitutions equally, failing to capture biological conservative swaps (e.g. L to I vs L to D).

2. **Embedding Similarity:**
   - **Concept:** Extracts the hidden states from the final layer of ESM-2 for the unmasked sequences. By mean-pooling these representations over the sequence (excluding special tokens like `<cls>` and `<eos>`), we get a 1D vector representing the structural/functional context of the protein. The cosine similarity between the wild-type and mutant embeddings measures how closely the mutant aligns with the original biological space.
   - **Significance:** A cosine similarity score close to 1.0 (e.g., > 0.99) indicates that the overall fold and biochemical properties are preserved.

> **Student Note on Embedding Vectors:**
> ESM-2 is trained on millions of natural protein sequences. Inside the model, it converts each amino acid token into a vector capturing its context—its likelihood of being near other specific residues, secondary structure context, and chemical properties. Averaging these token vectors yields a 1,280-dimensional representation of the entire sequence. Cosine similarity measures the angle between these vectors; if they point in almost the same direction, the model's structural understanding of the mutated protein is virtually identical to the wild-type.

</details>



### Implementation Prompt

**Prompt:**
> How can I check whether my optimized sequence is still
> functionally similar to the wild-type? I'd like to measure
> both simple sequence distance and a deeper similarity using
> the ESM-2 embeddings. Can you write code to compare the
> wild-type and optimized single-chain insulin sequences?


```python
# @title Code for computing similarity between two proteins
import numpy as np

def edit_distance(seq1: str, seq2: str) -> int:
  """Calculates the Levenshtein distance between two sequences using dynamic programming."""
  m, n = len(seq1), len(seq2)
  dp = np.zeros((m + 1, n + 1), dtype=int)

  # Initialize base cases
  for i in range(m + 1):
    dp[i, 0] = i
  for j in range(n + 1):
    dp[0, j] = j

  # Fill DP table
  for i in range(1, m + 1):
    for j in range(1, n + 1):
      cost = 0 if seq1[i - 1] == seq2[j - 1] else 1
      dp[i, j] = min(
          dp[i - 1, j] + 1,      # Deletion
          dp[i, j - 1] + 1,      # Insertion
          dp[i - 1, j - 1] + cost # Substitution
      )
  return int(dp[m, n])

def get_mean_pooled_embedding(sequence: str, model, tokenizer, device) -> torch.Tensor:
  """Extracts the mean-pooled embedding for a single sequence, excluding special tokens."""
  inputs = tokenizer(sequence, return_tensors="pt")
  inputs = {k: v.to(device) for k, v in inputs.items()}

  with torch.no_grad():
    # Pass output_hidden_states=True to get final layer representations
    outputs = model(**inputs, output_hidden_states=True)

  # Outputs contain hidden_states if requested
  last_hidden_state = outputs.hidden_states[-1]  # [1, seq_len + 2, hidden_dim]

  attention_mask = inputs["attention_mask"]
  seq_len = int(attention_mask.sum().item())

  # Extract representations of the actual amino acids (excluding <cls> at index 0 and <eos> at seq_len - 1)
  valid_embeddings = last_hidden_state[0, 1:seq_len-1, :]  # [seq_len, hidden_dim]

  # Mean pooling over the sequence length dimension
  mean_embedding = torch.mean(valid_embeddings, dim=0)  # [hidden_dim]
  return mean_embedding

def embedding_similarity(seq1: str, seq2: str, model, tokenizer, device) -> float:
  """Calculates the cosine similarity between the mean-pooled embeddings of two sequences."""
  emb1 = get_mean_pooled_embedding(seq1, model, tokenizer, device)
  emb2 = get_mean_pooled_embedding(seq2, model, tokenizer, device)

  # Calculate cosine similarity
  cos_sim = torch.nn.functional.cosine_similarity(emb1.unsqueeze(0), emb2.unsqueeze(0))
  return float(cos_sim.item())

# Compare WT and the optimized sequence from the optimization loop
distance_wt_opt = edit_distance(wt_sequence, optimized_sequence)
similarity_wt_opt = embedding_similarity(wt_sequence, optimized_sequence, model, tokenizer, device)

print(f"Comparison: Wild-Type vs Optimized Sequence")
print("-" * 50)
print(f"Edit Distance: {distance_wt_opt}")
print(f"ESM-2 Embedding Cosine Similarity: {similarity_wt_opt:.6f}")
```

When using AI to engineer a better, more stable protein (like a therapeutic drug), the algorithm must carefully balance two massive biological forces: **Hydrophobicity** and **Immunogenicity**. Here is what they are and why they dictate the success or failure of our engineered sequence.

#### 1. Hydrophobicity (The Folding Glue)
Hydrophobicity literally means "water-fearing." Because proteins exist in the watery environment of our blood and cells, amino acids that are hydrophobic behave exactly like oil in water—they aggressively repel it.
*   **Why it matters here:** Hydrophobicity is the primary engine of protein stability. To escape the surrounding water, all the "oily" hydrophobic amino acids clump together in the very center of the protein, folding the chain into a dense, solid core.
*   **The Engineering Goal:** If our optimization loop mutates a loose amino acid in the core into a bulkier, highly hydrophobic one (like Leucine or Isoleucine), it packs the protein core even tighter. #### 2. Immunogenicity (The Immune System Alarm)
While the *inside* of the protein handles stability, the *surface* handles the immune system. **Immunogenicity** is the measure of how strongly the human body will "freak out" and attack our newly engineered shape.
*   **The Quality Control Mechanism:** The human immune system constantly chops up floating proteins into tiny strings and inspects them. If the string exactly matches the wild-type human proteins you were born with, the immune system ignores it (it recognizes it as safe "self").
*   **Why it matters here:** If our AI mutates the protein to increase heat resistance, it might accidentally create a 3D shape the body has never seen before. The immune system will flag this mutated shape as an alien pathogen (like a virus) and sound a massive inflammatory alarm.

#### The Context of this Notebook: Why We Must Co-Optimize
If a therapeutic drug is highly immunogenic, the body will produce Anti-Drug Antibodies (ADAs) that instantly hunt down and destroy the drug every time it's injected, rendering it useless. Worse, it could trigger severe allergic reactions or dangerous autoimmune diseases.

This is why the multi-objective optimization loop in this notebook is so critical. We cannot just tell the AI to "make it as heat-resistant as possible," because it might mutate the surface beyond recognition to achieve that. By tracking both metrics simultaneously, the AI acts as a digital biochemist—searching for clever **hydrophobic** mutations that lock the protein together against heat, while verifying via the **immunogenicity** score that the protein still looks perfectly normal and safe to passing immune cells!

```python
# @title Code for defining a composite score
import pandas as pd

def calculate_composite_score(
    thermal: float,
    esm: float,
    immunogenicity: float,
) -> float:
  """Calculates the three-term composite score for multi-objective optimization.

  Components:
    1. Thermal stability (primary objective, in °C)
    2. ESM-2 evolutionary fitness (weighted by 2.0)
    3. Immunogenicity penalty (20.0× if presentation score > 0.5)
  """
  # 1. Thermal stability is our primary objective
  score = thermal

  # 2. ESM score (log-likelihood); higher is better
  score += esm * 2.0

  # 3. Penalize high immunogenicity
  if immunogenicity > 0.5:
    score -= (immunogenicity * 20.0)

  return score
```
